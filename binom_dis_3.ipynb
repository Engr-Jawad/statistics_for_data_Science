## binomial distribution 
### A manufacturing company produces electronic components. According to its quality-control records, 8% of all components are defective.
#### A quality inspector randomly selects 15 components from the production line.
1. What is the probability that exactly 3 components are defective?

```python
from scipy.stats import binom
n = 15
k = 3
def_probablity = 0.08
probablity = binom.pmf(k,n,def_probablity)
print("probablity that exactly three components are defective : ",probablity)
```

```python
# What is the probability that at least 2 components are defective?
probablity_of_2 = 1 - binom.cdf(1,n,def_probablity)
print("probablity that at least  2 components are defective : ",probablity_of_2)
```

```python
# What is the probability that at most 2 components are defective?
probablity_at_most_2 = binom.cdf(2,n,def_probablity)
print("probablity that at most two components are defective :  ",probablity_at_most_2)
```

```python
# What is the probability that between 1 and 4 components are defective (inclusive)?
probablity_betweeb_1_and_4 = binom.cdf(4,n,def_probablity) - binom.pmf(0,n,def_probablity)
print("the probablity between 1 and 4 : ",probablity_betweeb_1_and_4)
```

```python
# What is the probability that none of the components are defective?

probability_none_defective = binom.pmf(0, n, def_probablity)

print("Probability that none of the components are defective:",probability_none_defective)
```
