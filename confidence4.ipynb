# A university wants to estimate the average number of hours students study per day.

# A random sample of 25 students is selected, and the following results are obtained:

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import t
```

```python
sample_mean = 6.5 
sample_Std = 1.5
sample_size = 25
confidence_level = 0.95
alpha = 1 - confidence_level
degree_of_freedom = sample_size -1
print("Sample mean : ",sample_mean)
print("sample size : ",sample_size)
print("confidence level : ",confidence_level)
print("significance level : ",alpha)
print("degree_of_freedom : ",degree_of_freedom)
```

```python
standard_error = sample_Std / np.sqrt(sample_size)
print("Standard error : ",standard_error)
```

```python
critical_value = t.ppf(1-alpha/2,degree_of_freedom)
print("critical_value : ",critical_value)
```

```python
margin_error = critical_value * standard_error
print("margin_error : ",margin_error)
```

```python
lower_bound = sample_mean - margin_error
upper_bound = sample_mean + margin_error
print("lower bound : ",lower_bound)
print("upper_bound : ",upper_bound)
print("Conclusion : we are 95 % confident the average study time of student per day is between 5 to 7 hours .")
```
