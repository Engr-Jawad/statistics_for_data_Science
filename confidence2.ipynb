# A company manufactures laptop batteries. You collect a sample of 25 batteries.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import t
```

```python
Sample_mean= 8.5 
Sample_standard_deviation= 1.2
Sample_size= 25
Confidence_level = 0.95
alpha = 0.05
dof = Sample_size - 1
print("sample mean : ",Sample_mean)
print("Sample Standard deviation : ",Sample_standard_deviation)
print("Sample size : ",Sample_size)
print("confidence level  : ",Confidence_level)
print("Significance level : ",alpha)
```

```python
standard_error = Sample_standard_deviation / np.sqrt(Sample_size)
print("Standard error : ",standard_error)
```

```python
critical_value = t.ppf(1-alpha/2,dof)
margin_error = critical_value * standard_error
print("critical value : ",critical_value)
print("Margin error : ",margin_error)
```

```python
lower_limit = Sample_mean - margin_error
upper_limit = Sample_mean + margin_error
print("lower limit : ",lower_limit)
print("upper limit : ",upper_limit)
print("Conclusion : i am 95 % confident that the average battery life is between 8.00 to 8.99999.")
```
