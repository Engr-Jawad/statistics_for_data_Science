# A company claims that its employees complete a task in an average of 30 minutes.

# A researcher collects data from 12 employees:

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import t,ttest_1samp
```

```python
completion_time = [32, 29, 35, 31, 28, 34,
                   33, 30, 36, 29, 32, 31]
df = pd.DataFrame({"completion_time" : completion_time})
df
```

```python
sample_mean = df["completion_time"].mean()
sample_size = len(df)
Degree_of_freedom = sample_size - 1 
population_mean = 30
alpha = 0.05
print("Sample mean : ",sample_mean)
print("Sample size : ",sample_size)
print("Degree_of_freedom : ",Degree_of_freedom)
print("Population mean : ",population_mean)
print("significance level : ",alpha)
```

# t test stats

```python
t_score , p_value = ttest_1samp(df["completion_time"],population_mean)
print("t test statistics : ",t_score.round(4))
print("p value : ",p_value.round(4))
```

```python
# critical value 
critical_value = t.ppf(1-alpha/2 , Degree_of_freedom)
lower_fence = -critical_value
upper_fence = critical_value
print("Critical value : ",critical_value)
print("lower fence : ",lower_fence)
print("upper fence : ",upper_fence)

```

```python
if p_value < alpha :
    print("Decision : we reject the null hypothesis\n""Conclusion : there is sufficient evidence at 5% significance level .")
else:
    print("Decision : we are fail to reject the null hypothesis \n""Conclusion : there is not sufficent evidence at 5 % significance level .")
```

```python
x = np.linspace(-4,4,1000)
y = t.pdf(x,Degree_of_freedom)
plt.Figure(figsize=(8,9))
plt.plot(x,y,label = "t distribution")
plt.axvline(critical_value,color = "Red",linestyle = "--",label = "critical_value")
plt.axvline(lower_fence,linestyle = "--",color="Red")
plt.axvline(t_score,color = "green",label = "t test stats")
plt.title("One-Sample T-Test Distribution")
plt.xlabel("T-statistic")
plt.ylabel("Probability Density")
plt.legend()
plt.grid()
plt.show()
```
