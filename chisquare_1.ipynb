# Student Programming Language Preferences

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import chisquare
```

```python
observed_frequency = np.array([80,50,30,40])
programming_language = ["python","java","c++","javascript"]
df = pd.DataFrame({"programming_lang":programming_language,"Ratio":observed_frequency})
df
```

```python
alpha = 0.05
confidence_level = 0.95

total_observed = np.sum(df["Ratio"])
print("total observation : ",total_observed)

k = len(df)
print("length of the data frame : ",k)

expected = np.full(k,total_observed/k)
print("expected :",expected)

chi_square_stats,p_value = chisquare(observed_frequency,expected)
print(chi_square_stats,"\n",p_value)
```

```python
if p_value < alpha :
    print("Decision :we reject the null hypothesis")
    print("Conclusion : there are sufficient evidence at 95 % confidence level that the student prefere the progarmming language at diffrent ratio . ")
else:
    print("Decision : we are fail to reject the null hypothesis .")
    print("Conclusion : there are sufficient evidence at 95 % confidence leve that the student prefer the programming language equally .")
```

```python
x = np.arange(len(df))
plt.figure(figsize=(8,9))
width = 0.35
plt.bar(x - width/2,observed_frequency,width,label = "observed frequency")
plt.bar(x + width/2 , expected,width,label = "expected Frequecy")
plt.title("Observed Vs Expected Frequency")
plt.xlabel("programming language")
plt.ylabel("Number of students ")
plt.legend()

plt.show()
```
