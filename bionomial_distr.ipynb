# Binomial distribution
formula  ---->  p(X=k)= nCk(p)power k (p)power n-k

```python
# Question 1
# A fair coin is tossed 6 times.
# Find the probability of getting exactly 2 heads.

from scipy.stats import binom
n=6
p=1/2
probablity = binom.pmf(2,n,p)
print("probablity:",probablity)
```

```python
#1. Coin Toss
# A fair coin is tossed 7 times.
# What is the probability of getting exactly 4 heads?
n=7
p=1/2
probablity_of_4 = binom.pmf(4,n,p)
print("probablity of getting exactly four : ", probablity_of_4)
```

```python
# 2. Die Roll
# A die is rolled 10 times.
# What is the probability of getting exactly 2 sixes?
n=10
p=1/6
probablity_of_6 = binom.pmf(2,n,p)
print("probablity of getting exactly 2 sixes : ", probablity_of_6)
```
