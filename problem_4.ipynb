```python
data = [21,23,23,43,56,77,55,65,54,98,78,87,88,18,98]
vare = np.var(data,ddof=1)  # mean sample varience
vare2 = np.var(data,ddof= 0) # mean population vareince
print(vare)
print(vare2)
```

# now calculating varience manually 


```python

def vareince(data):
    n = len(data)
    mean = sum(data)/n
    deviation = [(x - mean) ** 2 for x in data]
    varience_value = sum(deviation)/n
    return varience_value


print(vareince(data))
```
