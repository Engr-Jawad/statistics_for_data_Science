# implementing Z test practically

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt 
from scipy.stats import norm
```

```python
np.random.seed(42)
crop_yeild_observation = np.random.normal(loc = 40 , scale = 6 , size = 36)
df = pd.DataFrame({"crop_yeild_observation" : crop_yeild_observation})
df.head()
```

```python
sample_mean = df["crop_yeild_observation"].mean()
sample_size = len(df)
population_mean = 40
population_Std = 6
alpha = 0.05
print("Sample mean : ",sample_mean)
print("Sample size : ",sample_size)
print("Population mean : ",population_mean)
print("population standard deviation : ",population_Std)
print("significane value : ",alpha)
```

# Calculating Z test Stats

```python
standard_error = population_Std / np.sqrt(sample_size)
z_score = (sample_mean - population_mean)/standard_error
print("standard error : ",standard_error)
print("Z Score  : ",z_score.round(3))
```

```python
critical_value = norm.ppf(alpha)
lower_fence = -critical_value
upper_fence = critical_value
print("critical value : ",critical_value)
print("lower_fence : ",lower_fence)
print("upper_fence : ",upper_fence)
```

```python
p_value = norm.cdf((z_score))
print("P value : ",p_value)
```

```python
if p_value < alpha:
    print("we reject the null hypothesis")
else:
    print("we are fail to reject the null hypothesis")
```

```python
x = np.linspace(-4,4,1000)
y = norm.pdf(x)
plt.plot(x,y,color = "green",label = "Standard normal distribution")
plt.title("Standard Normal Distribution")
plt.xlabel("Z score")
plt.ylabel("probablity denstiy")
plt.grid(True)
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(9,7))
plt.plot(x,y,color = "green",label = "standard normal distribution")
plt.axvline(critical_value,linestyle = "--",label = "Critical value (-1.6444)")
plt.title("Standard normal distribution (z test) left taild")
plt.xlabel("z_score")
plt.ylabel("probablity density")
plt.legend()
plt.grid(True)
plt.show()
```
