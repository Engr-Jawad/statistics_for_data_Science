# log normal distribution


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
house_prices = [
    120000, 150000, 180000, 200000, 220000,
    250000, 280000, 300000, 350000, 450000,
    600000, 900000, 1500000
]
```

```python
df = pd.DataFrame({"house price":house_prices})
print(df.head())
# mean
mean= df.mean()
print("mean :\n", mean)
# median
median = df.median()
print("median \n ",median)
# standard deviation
std = np.std(df)
print("standard deviation : ", std)
```

```python
plt.figure(figsize=(8,7))
plt.hist(df["house price"],bins=5,color="blue",edgecolor="black")
plt.title("distribution of the house prices")
plt.xlabel("house price")
plt.ylabel("frequency")
plt.show()
```

```python
df["log_house_price"] = np.log(df["house price"])
print(df)
mean_log_house_price = np.mean(df["log_house_price"])
print("mean of the log house price : ",mean_log_house_price.round(3))
median_of_log_house_price = np.median(df["log_house_price"])
print("median of the log house price : ",median_of_log_house_price.round(3))
std_of_log_house = np.std(df["log_house_price"])
print("standard deviation of the log house price : ", std_of_log_house.round(3))
```

```python
fig,ax = plt.subplots(1,2,figsize=(8,7))
# subplot one 
ax[0].hist(df["house price"],bins=8,edgecolor = "black")
ax[0].set_title("orignal distribution")
ax[0].set_xlabel("house price")
ax[0].set_ylabel("frequency")



# subplot two
ax[1].hist(df["log_house_price"],bins=8,color="purple",edgecolor="black")
ax[1].set_title("after tranformation distribution")
ax[1].set_xlabel("house price")
ax[1].set_ylabel("frequency")

plt.tight_layout()
plt.savefig("log_normal.png",dpi=200,bbox_inches="tight")
plt.show()
```
