# implementing z test practically 

```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)
scores = np.random.normal(loc = 73 , scale = 10 , size= 50)
df = pd.DataFrame({"Scores_Of_Student":scores})
print(df.head(3))
```

```python
sample_mean = df["Scores_Of_Student"].mean()
sample_size = len(df)
population_mean = 70
population_std = 10
alpha = 0.05

print("Sample Mean : ", sample_mean)
print("Sample Size : ",sample_size)
print("Population Mean : ",population_mean)
print("Population standard deviation : ",population_std)

```

# Z test stat

```python
from scipy.stats import norm
standard_error = population_std / np.sqrt(sample_size)
z_test = (sample_mean - population_mean)/ standard_error
print("Standard error : ", standard_error.round(3))
print("Z Test Stat : ",z_test.round(3))
```

```python
# Because this is a two-tailed test, divide the significance level by two.
critical_value = norm.ppf(1-alpha / 2 )
lower_fence = -critical_value
upper_fence = critical_value
print("Critical Value : ",critical_value.round(3))
print("lower_fence : ",lower_fence.round(3))
print("upper_fence : ",upper_fence.round(3))
```

```python

p_value = 2 * norm.sf(abs(z_test))
print("P Value : ",p_value.round(3))
```

```python
if p_value < alpha:
    print("Reject the null hypothesis.")
else:
    print("Fail to reject the null hypothesis.")
```
