# calculas 

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(10,100,1000000)
y = np.sqrt(x)
plt.figure(figsize = (8,9))
plt.plot(x,y,label=" Y = /x")
plt.title("Root of x")
plt.xlabel("x")
plt.ylabel("y")
plt.grid()
plt.legend()
plt.show()
```

# Sin function graph

```python
x = np.linspace(-2*np.pi , 2*np.pi,1000)
y = np.sin(x)
plt.figure(figsize=(8,6))
plt.plot(x,y)
plt.axvline(0,color="black",linewidth=2,linestyle="--",label = "y axis")
plt.axhline(0,color="black",linestyle="--",linewidth=2,label="x axis")

plt.title("Sin function ")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid()
plt.show()

```
