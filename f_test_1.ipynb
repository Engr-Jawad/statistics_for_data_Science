# A researcher wants to compare the variability of exam scores from two classes.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import f
```

```python
n1 = 10
n2 = 12
var1 = 25
var2 = 9
alpha = 0.05
confidence_level = 0.95
F_test_stat = var1 / var2
df1 = n1-1
df2 = n2-1
lower_critical_value = f.ppf(alpha/2 , df1,df2)
upper_critical_value = f.ppf(1 - alpha / 2 ,df1,df2)

# Two-tailed p-value
cdf_value = f.cdf(F_test_stat, df1, df2)
p_value = 2 * min(cdf_value, 1 - cdf_value)

print("F-statistic:", F_test_stat)
print("DF numerator:", df1)
print("DF denominator:", df2)
print("Lower critical value:", lower_critical_value)
print("Upper critical value:", upper_critical_value)
print("P-value:", p_value)

if p_value < alpha:
    print("Decision : we are reject the null hypothesis .")
    print("Conclusion : there are sufficient statistical evidence at the 5 % significane level to conclude that the population variances differ .")
else:
    print("Decision : we are fail to reject the null hypothesis .")
    print("Conclusion : There is insufficient statistical evidence at the 5% significance level to conclude that the population variances differ.")
```

```python
x = np.linspace(0,8,1000)
y = f.pdf(x,df1,df2)
plt.plot(x,y,label="F distribution")
plt.axvline(lower_critical_value,linestyle = "--",label ="lower critical value ",color = "green")
plt.axvline(upper_critical_value,linestyle = "--",label = "upper critical value ",color = "green")
plt.axvline(F_test_stat,linestyle = "solid",label =" f test stats",color = "red")
plt.title("compare the variability of exam scores from two classes")
plt.xlabel("F value ")
plt.ylabel("probablity density")
plt.legend()
plt.tight_layout()
plt.show()
```
