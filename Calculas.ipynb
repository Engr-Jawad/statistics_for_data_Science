# calculas 

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(10,100,1000000)
y = np.sqrt(x)
plt.figure(figsize = (8,9))
plt.plot(x,y,label=" Y = /x")
plt.title("Root of x")
plt.xlabel("x")
plt.ylabel("y")
plt.grid()
plt.legend()
plt.show()
```

# Sin function graph

```python
x = np.linspace(-2*np.pi , 2*np.pi,1000)
y = np.sin(x)
plt.figure(figsize=(8,6))
plt.plot(x,y)
plt.axvline(0,color="black",linewidth=2,linestyle="--",label = "y axis")
plt.axhline(0,color="black",linestyle="--",linewidth=2,label="x axis")

plt.title("Sin function ")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid()
plt.show()

```

# Linear Function (Straight Line)

```python
# Function: f(x)=2x+1
x = np.linspace(2,10,10000)
y = 2 * x + 1 
plt.figure(figsize=(8,6))
plt.plot(x,y,color = "green",linewidth =1.5,label = "liner function")
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.legend()
plt.show()
```

#  Quadratic Function 

```python
# Function: f(x)=x**2
x = np.linspace(5,10,100000)
y = x **2
plt.figure(figsize=(8,6))
plt.plot(x,y,label = "Quadratic function")
plt.title("Quadratic function")
plt.xlabel("X axis")
plt.ylabel("y axis")
plt.legend()
plt.show()
```

# Exponential Function

```python
# Function: f(x)= e power x
x = np.linspace(1,10,10000)
y = np.exp(x)
plt.figure(figsize=(8,7))
plt.plot(x,y,color="red",label="exponential function")
plt.title("Exponential fucntion")
plt.xlabel("X axis")
plt.ylabel("y axis")
plt.legend()
plt.show()
  
```
