# Problem on the binomial distribution 
### A email company has 20 % click through rate ? the email is send to 12 users . what is the probablity that emails is clicked 3 users exactly .


```python
from scipy.stats import binom
```

```python
n = 12 
k = 3 
p = 0.20
probablity = binom.pmf(k,n,p)
print("probablity of eamils that clicked 3 users exactly : ",probablity*100)
```
