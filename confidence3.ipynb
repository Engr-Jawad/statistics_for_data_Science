# An agricultural researcher wants to estimate the average crop yield.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import norm
```

```python
sample_mean = 42 
population_Std = 8
sample_size = 64
confidence_level = 0.99
alpha = 0.01
print("Sample mean : ",sample_mean)
print("population_Std : ",population_Std)
print("sample size : ",sample_size)
print("confidence level : ",confidence_level)
print("Significance value : ",alpha)
```

```python
standard_error = population_Std / np.sqrt(sample_size)
print("standard error : ",standard_error)
```

```python
critical_value = norm.ppf(1-alpha/2)
print("critical value : ",critical_value)
```

```python
margin_error = critical_value * standard_error
print("margin error : ",margin_error)
```

```python
lower_limit = sample_mean - margin_error
upper_limit = sample_mean + margin_error
print("lower limit : ",lower_limit)
print("upper limit : ",upper_limit)
print("conclusion : i am 99 % confident that the average crop tons between 39 to 44 .")
```
