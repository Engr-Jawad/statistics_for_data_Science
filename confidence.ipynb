# A university wants to estimate the average marks of all students.

```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
from scipy.stats import norm
```

```python
sample_size = 36 
sample_mean = 75
confidenc_interval = 0.95
population_Std = 12
alpha = 0.05
print("sample size : ",sample_size)
print("sample mean : ",sample_mean)
print("significane level : ",alpha)
print("population std : ",population_Std)
print("Confidence interval : ",confidenc_interval)
```

```python
standard_error = population_Std / np.sqrt(sample_size)
print("Standard error : ",standard_error)
```

```python
critical_value = norm.ppf(1-alpha/2)
print("critical value : ",critical_value)
```

```python
margin_error = critical_value * standard_error
print("Margin error : ",margin_error)
```

```python
lower_conf_interval = sample_mean - margin_error
upper_conf_interval = sample_mean + margin_error
print("lower confidence interval : ",lower_conf_interval)
print("upper confidence interval : ",upper_conf_interval)
print("conclusion : i am 95 % confident the average marks of the student is between 71 to 78 .")
```
