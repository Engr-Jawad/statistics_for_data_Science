# implementing z test practically 

```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
from scipy.stats import norm
```

```python
np.random.seed(42)
satisfication_score = np.random.normal(loc = 7.6,scale = 1.5 ,size = 100)
df = pd.DataFrame({"Satisfication_Score":satisfication_score})
df.head()
```

```python
sample_mean = df["Satisfication_Score"].mean()
print("Sample mean : ",sample_mean.round(4))
sample_size = len(df)
print("Sample Size : ",sample_size)
population_mean = 7.6
population_std = 1.5
alpha = 0.02
print("population mean : ",population_mean)
print("population Std : ",population_std)
print("Significance value : ",alpha)
```

# creating function for the calculating the z test score

```python
def z_test(sample_mean,sample_size,population_std,population_mean,alpha,alternative):
    standard_Error = population_std/np.sqrt(sample_size)
    z_test_score = (sample_mean - population_mean)/standard_Error
     # Critical value and p-value
    if alternative == "two-sided":
        crit = norm.ppf(1 - alpha / 2)
        p_value = 2 * norm.sf(abs(z_test_score))
    elif alternative == "less":
        crit = norm.ppf(alpha)
        p_value = norm.cdf(z_test_score)
    else:  # "greater"
        crit = norm.ppf(1 - alpha)
        p_value = norm.sf(z_test_score)

    # Decision
    if p_value < alpha:
        decision = "Reject H0"
    else:
        decision = "Fail to reject H0"

    # Results
    print(f"--- {alternative} test ---")
    print(f"Z-statistic    : {z_test_score:.3f}")
    print(f"Critical value : {crit:.3f}")
    print(f"P-value        : {p_value:.4f}")
    print(f"Decision       : {decision}\n")
    
```

```python
z_test(sample_mean=7.6, population_mean=8, population_std=1.5, sample_size=100, alpha=0.02, alternative="two-sided")
```
