## binomial distribution 
### A manufacturing company produces electronic components. According to its quality-control records, 8% of all components are defective.
#### A quality inspector randomly selects 15 components from the production line.
1. What is the probability that exactly 3 components are defective?

```python
from scipy.stats import binom
n = 15
k = 3
def_probablity = 0.08
probablity = binom.pmf(k,n,def_probablity)
print("probablity that exactly three components are defective : ",probablity)
```

```python
# What is the probability that at least 2 components are defective?
k = 2 
probablity_of_2 = binom.pmf(k,n,def_probablity)
print("probablity that exactly 2 components are defective : ",probablity_of_2)
```
