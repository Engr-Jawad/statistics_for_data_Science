# Comparing two population variances

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import f 
```

```python
n1 = 15 
n2 = 12
var1 = 64 
var2 = 25 
alpha = 0.01
confidence_level = 0.99
dof1 = n1-1
dof2 = n2 - 1 
print("sample size of machine 1 : ",n1)
print("sample size of machine 2 : ",n2 )
print("significance level : ",alpha)
print("confidence level : ",confidence_level)
print("varience of machine 1 : ",var1)
print("varience of machine 2 : ",var2)
print("degree of freedom of sample 1 : ",dof1)
print("degree of freedom of sample 2 : ",dof2)

f_stats = var1 / var2
print("F test stats : ",f_stats)

upper_critical_value = f.ppf(1 - alpha /2 ,dof1,dof2)
lower_critical_value =  f.ppf(alpha / 2, dof1, dof2)
print("critical value : ",upper_critical_value)
print("lower critical value : ",lower_critical_value)

p_value =2 *  f.sf(1 - alpha / 2 ,dof1,dof2)
print("p value : ",p_value)
```

```python
if p_value < alpha :
    print("Decision : we reject the null hypothesis .")
    print("Conclusion : there are sufficient evidence at 1  % significance level that the machine 1 is different from the machine 2 . ")
else :
    print("Decision : we are fail to reject the null hypothesis .")
    print("Conclusion : there are sufficient evidence at  1 % significance level that the machine 1 and machine 2 vareince are same .")
```
