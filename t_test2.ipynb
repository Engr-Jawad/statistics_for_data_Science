# A university claims that the average study time of its students is 6 hours per day.

# A researcher selects 10 students and records their daily study hours:

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import ttest_1samp
```

```python
study_hours = [5.5, 6.2, 5.8, 4.9, 6.5,
               5.7, 6.1, 5.3, 5.9, 6.0]

df = pd.DataFrame({"study_hours":study_hours})
df
```

```python
population_mean = 6
sample_size = len(df)
alpha = 0.05
sample_mean = df["study_hours"].mean()
print("population mean : ",population_mean)
print("sample size : ",sample_size)
print("significance level : ",alpha)
print("sample_mean : ",sample_mean)
```

# t test stat

```python
t_stats,p_value = ttest_1samp(df["study_hours"],population_mean)
print("t test statistics score : ",t_stats.round(4))
print("p value : ",p_value.round(4))
```

```python
if p_value < alpha:
    print("Decision : we reject the null hypothesis." "\n" "Conclusion :  there is suffiecent evidence at 5 % significance level that the null hypothesis is different .")
else:
    print("Decision : we are fail to reject the null hypothesis . ""\n" "Conclusion : there is not sufficent evidence at 5 % significance level .")
```

# visualizing the t test stats

```python
plt.figure(figsize=(8,9))
plt.hist(df["study_hours"],bins=6,color = "green",edgecolor = "Black")
plt.axvline(population_mean,color = "blue",label = "blue line (population mean)")
plt.axvline(sample_mean,color="red",label = "red line (sample mean)")
plt.title("Distribution of Student study hours")
plt.xlabel("study hours")
plt.ylabel("frequency")
plt.legend()
plt.show()
```

```python
from scipy.stats import t,ttest_1samp
# Parameters
df_value = 9
alpha = 0.05
t_stats = -1.427

# Critical values
critical_value = t.ppf(1 - alpha/2, df_value)
print(critical_value)

# Generate x values
x = np.linspace(-4, 4, 500)

# Calculate probability density
y = t.pdf(x, df_value)

# Plot distribution
plt.figure(figsize=(10, 6))
plt.plot(x, y, label="T-distribution")

# Critical values
plt.axvline(-critical_value, color="red",
            linestyle="--", label="Critical values")
plt.axvline(critical_value, color="red",
            linestyle="--")

# Test statistic
plt.axvline(t_stats, color="green",
            linestyle="-", label="T-statistic")

plt.title("One-Sample T-Test Distribution")
plt.xlabel("T-statistic")
plt.ylabel("Probability Density")
plt.legend()
plt.grid()
plt.show()
```
