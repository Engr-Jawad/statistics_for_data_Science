# Gussian or Normal distribuiton


```python
import numpy as np
import matplotlib.pyplot as plt
from  scipy.stats import norm
```

```python
mu = 70
sigma = 10
x= np.linspace(30,110,1000)
y = norm.pdf(x,mu,sigma)
plt.plot(x,y,color="black",linewidth=2,label = "Normal distiribution")
x1 = np.linspace(mu - sigma , mu + sigma,1000)
plt.fill_between(x1,norm.pdf(x1,mu,sigma),color="yellow",alpha=0.3,label = "68 % region +-1alpha")
x2 = np.linspace(mu-2*sigma,mu+2*sigma,1000)
plt.fill_between(x2,norm.pdf(x2,mu,sigma),color="green",alpha=0.2,label="95.8 % region +-2 alpha")
x3 = np.linspace(mu-3*sigma,mu+3*sigma,1000)
plt.fill_between(x3,norm.pdf(x3,mu,sigma),color="red",alpha=0.1,label = "99.7% region +-3 alpha")
plt.title("Normal Distribution")
plt.xlabel("Score")
plt.ylabel("probablity")
plt.legend()
plt.show()
```

# A bank analyzes the monthly transaction amounts of its customers. Assume the transaction amounts follow a normal distribution:

* Mean μ=50,000
* Standard deviation σ=10,000

* Answer the following:

* What is the probability that a customer's transaction amount is less than 60,000?
* What is the probability that it is between 40,000 and 60,000?
* What is the probability that it is greater than 70,000?
* Generate 10,000 random transaction amounts and visualize their distribution.

```python
mean = 50000
std = 10000
#Generate random data
data = np.random.normal(
    loc=mean,
    scale=std,
    size=10000
)
# probablity of a customers transication less than  60000 .
p_less_60000 = norm.cdf(60000,mean,std)
print("probablity of  customers transiction less than 60000 = ",p_less_60000.round(3))

# probablity b/w 40000 and 60000
p_between = norm.cdf(60000,mean,std) - norm.cdf(40000,mean,std)
print("probablity of customer transiction b/w 40000 and 60000 = ",p_between.round(3))

# probablity greater than 700000
p_greater_700000 = 1-norm.cdf(70000,mean,std)
print("probablity of customer transiction greater than 70000 = ",p_greater_700000.round(3))


# Plot histogram
plt.hist(data, bins=30, edgecolor="black")

plt.title("Normal Distribution of Transaction Amount")
plt.xlabel("Transaction Amount")
plt.ylabel("Frequency")

plt.show()
```
